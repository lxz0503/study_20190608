```python
from matplotlib import pyplot as plt
import pandas as pd
import numpy as np

# 显示中文 
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 准备画布
plt.figure('pie', figsize=(20,8), dpi=100,facecolor='lightgray')

# 画图准备
labels = ['python', 'C', 'C++', 'JS', 'PHP', 'JAVA']   # 饼图各个饼块
sizes = [2, 5, 12, 70, 2, 9]   # 总数是100，随便一个就行   各个饼块的数量
explode = (0.1, 0, 0, 0.1, 0, 0)   # 0.1表示把那部分从饼图里拿出来
plt.pie(sizes, labels=labels, explode=explode, autopct='%.1f%%', shadow=False, startangle=150)

# 添加标题
plt.title('Pie',fontdict={'fontsize':20, 'color':'b'})
# plt.savefig('pie.jpg')

# 展示图片
plt.show()
```
