# 等间隔生成多少个数

```python
import numpy as np
np.linspace(0,100,11)
```

```python
np.arange(10,50,2)    # 不包括50   左闭右开
```

```python
np.logspace(0,2,3)   # 10的2次方生成3个数
```

```python
np.random.rand()   # 返回0-1之间的一个随机数
```

```python
np.random.rand(2,3)    # 生成 2*3的array   二维数组
```

```python
np.random.uniform(1,10,5)    # 均匀分布 随机采样，左闭右开  即不包括右边
```

```python
np.random.randint(1,10,5)   # 整数随机采样 
```

```python
np.random.uniform(low=1,high=10,size=(2,3))   # 生成一个2*3的二维数组  即array
```
