```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
import random

# 显示中文 
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

movie_name = ['雷神','变形金刚','肖申克的救赎','阿甘正传','星球大战']

# x轴
x = range(len(movie_name))
# y轴 票房数据
y = [1000, 2000, 5000, 500, 4000]

# 创建画布
plt.figure(figsize=(20,8), dpi=100)

# 绘制柱状图
plt.bar(x,y,width=0.5,color=['m','g','r','c','k'])

# 修改x轴的刻度显示---使其显示文字，否早默认必须是数字
plt.xticks(x, movie_name)

# 添加网格
plt.grid(True,color='red',ls='-.',lw=0.3)

# 添加标题 x y轴名称
plt.xlabel('电 影',fontdict={'fontsize':15,'color':'b'})
plt.ylabel('票房(万元)',fontdict={'fontsize':15,'color':'b'})
plt.title('各大电影票房', fontdict={'fontsize':20, 'color':'b'})


plt.show()
```
