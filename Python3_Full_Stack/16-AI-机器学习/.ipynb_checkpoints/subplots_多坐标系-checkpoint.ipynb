```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
import random

# 生成数据
x = range(60)
y_beijing = [random.uniform(10,15) for i in x]
y_shanghai = [random.uniform(20,25) for i in x]

# 显示中文 
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建画布---相当于 1*2 即 一行里显示两个图像
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(20,8),dpi=100)

# 画图
# plt.plot(x,y_beijing,'r*-',label='北京')        # 下面是r * - 各自代表的参数
axes[0].plot(x,y_beijing, label='北京', color='r', marker='*', linestyle='-') 
axes[1].plot(x,y_shanghai,'g.--',label='上海')   # 使用这种位置参数的话就必须把 color marker等参数放中间

# 添加x y轴刻度
y_ticks = range(40)
x_ticks_labels = ['11:{}分'.format(i) for i in x]

# 下面两个只是axes[0]和axes[1]的区别，其它都一样
axes[0].set_xticks(x[::5])
axes[0].set_yticks(y_ticks[::5])    # 切片
axes[0].set_xticklabels(x_ticks_labels[::5])   # 切片  每5个取一个数据

axes[1].set_xticks(x[::5])
axes[1].set_yticks(y_ticks[::5])
axes[1].set_xticklabels(x_ticks_labels[::5])

# 添加网格
axes[0].grid(True,color='red',ls='-.',lw=0.3)
axes[1].grid(True,color='g',ls='-.',lw=0.3)

# 添加x y轴描述信息和title
axes[0].set_xlabel('时 间',fontdict={'fontsize':15,'color':'b'})
axes[0].set_ylabel('温 度',fontdict={'fontsize':15,'color':'b'})
axes[0].set_title('北京一小时温度变化', fontdict={'fontsize':20, 'color':'b'})

axes[1].set_xlabel('时 间',fontdict={'fontsize':15,'color':'g'})
axes[1].set_ylabel('温 度',fontdict={'fontsize':15,'color':'g'})
axes[1].set_title('上海一小时温度变化', fontdict={'fontsize':20, 'color':'g'})

# 添加图例
axes[0].legend()
axes[1].legend()

#
plt.show()
```
