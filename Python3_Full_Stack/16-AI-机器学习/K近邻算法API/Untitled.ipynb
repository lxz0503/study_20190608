```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.neighbors import KNeighborsClassifier 
```

```python
# 构造简单数据
x = [[1],[2],[3],[0],[100]]
y = [1,1,1,0,1]

# 实例化API
estimator = KNeighborsClassifier(n_neighbors=5)

# 使用fit方法进行训练
estimator.fit(x,y)

# 预估其他值
res = estimator.predict([[8]])
```
