```python
import numpy as np

stock_change = np.random.normal(0,1,(8,10))   # 高斯分布，也是正态分布数据
stock_c = stock_change[0:5,0:5]    # 取前5行和前5列
stock_c
```

```python
stock_c[stock_c>1] = 2     # 把里面大于1的值赋值成2
stock_c
```

```python
temp = np.where(np.logical_and(stock_c>0.5,stock_c<1),1,0)   # 逻辑运算 满足大于0.5且小于1，才赋值为1，否则赋值为0
temp
```

```python
np.where(stock_c>1,2,0)   # 里面大于1的赋值为2，否则赋值为0
```

```python
stock_change.max(axis=1)    # 这里是按列求出最大值  这个按行或者按列都是变动的  在pandans里面axis=1不一定代表按列
```

```python
stock_change.argmax(axis=1)    # 按行 从左到右开始计数，每行最大值的下标
```

```python
# 矩阵乘法 求加权成绩 

a = np.array([[80,86], [90,98],[76,78],[90,75]])  # 4个人 两门成绩  4*2 矩阵
b = np.array([[0.7],[0.3]])   # 上面a成绩单中分别占的权重      2*1 矩阵
b
np.matmul(a,b)   # 矩阵相乘  
```

```python
b   # 2*1  即 2行1列
```

```python
np.dot(a,b)    # 也是矩阵相乘 
```

```python
np.dot(a,10)    # 矩阵a中每个元素都乘以10  即 点乘 
```
