## 加载数据

```python
import pandas as pd

df = pd.read_csv("东京奥运会奖牌数据.csv") 
#df = pd.read_html("https://olympics.com/tokyo-2020/olympic-games/zh/results/all-sports/medal-standings.htm")[0]
df.head()
```

## 5-1 数据修改



### 1- 数据修改｜列名



将原 `df` 列名 `Unnamed: 2`、`Unnamed: 3`、`Unnamed: 4` 修改为 `金牌数`、`银牌数`、`铜牌数`

```python
df.rename(columns={'Unnamed: 2':'金牌数','Unnamed: 3':'银牌数','Unnamed: 4':'铜牌数'},inplace=True)
df.head()
# ans(1)
```

### 2 - 数据修改｜行索引



将第一列（排名）设置为索引

```python
df.set_index('排名').head()
# 如果真要生效 就要添加参数 inplace=True
# df.set_index('排名',inplace=True).head()
```

### 3 - 数据修改｜修改索引名



修改索引名为 金牌排名

```python
df.rename_axis('金牌排名').head()
```

### 4 - 数据修改｜修改值



将 ROC（第一列第五行）修改为 俄奥委会

```python
df.iloc[4,1] = '俄奥委会'
df.head()
```

### 5 - 数据修改｜替换值（单值）



将金牌数列的数字 `0` 替换为 `无`

```python
df['Unnamed: 2'].replace(to_replace=0,value='无')
ans(5)
```

### 6 - <font color = '#FB8E00'>数据修改</font>｜替换值（多值）



同时替换

- 将 `无` 替换为 缺失值
- 将 0 替换为 `None`

注意：缺失值的 Nan 该怎么生成？

```python
ans(6)
# 多值替换
import numpy as np
df.replace(['无',0],[np.nan,'None'])
```

### 8 - 数据修改｜修改类型

### 9 - 数据增加｜新增列（固定值）



**重新加载数据** 并 **新增一列** 比赛地点，值为 `东京`


```python
df['比赛地点'] = '东京'
df.head()
```

### 10 - 数据增加｜新增列（计算值）



**新增一列** 金银牌总数列，值为该国家金银牌总数

```python
df['金银牌总数'] = df['金牌数'].fillna('0') + df['银牌数'].fillna('0')
df.head()
# ans(10)
```

### 11 -数据增加｜新增列（比较值）

新增一列 最多奖牌数量 列，值为该国 `金、银、铜` 牌数量中最多的一个奖牌数量

例如美国银牌最多，则为41，中国为38

```python
ans(11)
# 答案 分解 xiaozhan
# step 1  max(1)里面的参数1代表按行求最大值  只取3列数据
df['最多奖牌数量'] = df[["金牌数", "银牌数",'铜牌数']].max(axis=1)
df.head()
```

### 12 - 数据增加｜新增列（判断值）

新增一列 金牌大于30

如果一个国家的金牌数大于 30 则值为 是，反之为 否

```python
df['金牌大于30']  = np.where(df['金牌数'] > 30, '是', '否')
df.head()
# ans(12)
```

### 13 - 数据增加｜增加多列

新增两列，分别是

- 金铜牌总数（金牌数+铜牌数）
- 银铜牌总数（银牌数+铜牌数）

```python
ans(13)
df.assign(金铜牌总数=df.金牌数 + df.铜牌数,
         银铜牌总数=df.银牌数+df.铜牌数) 
```

### 14 - 数据增加｜新增列（引用变量）

新增一列金牌占比，为各国金牌数除以总金牌数（gold_sum）

```python
gold_sum = df['金牌数'].sum()
gold_sum
```

```python
# 方法1 xiaozhan
df['金牌占比'] = df['金牌数'] / df['金牌数'].sum()
# df['金牌占比'].apply(lambda x:format(x,'.2f'))
df['金牌占比'] = df['金牌占比'].apply(lambda x:format(x,'.2%'))
df['金牌占比']
# 方法2 也是作者的答案 不太理解第一行代码
# df.eval(f'金牌占比 = 金牌数 / {gold_sum}',inplace=True)
# df['金牌占比'].apply(lambda x:format(x,'.2%'))
```

###  43 - 筛选值｜query（引用变量）

使用 query 提取 金牌数 大于 金牌均值的国家

```python
ans(43)
# xiaozhan 下面这种用法需要学习
gold_mean = df['金牌数'].mean()
df.query(f'金牌数 > {gold_mean}')
```

```python
# xiaozhan  方法2
df['金牌数'] = df['金牌数'].apply(lambda x:x > df['金牌数'].mean())
# df['金牌数']
df[df['金牌数']]
df[df['金牌数']]['国家奥委会'] # 一个列表
```

```python
# pandas中 query的用法举例 其实就是对列进行查询 xiaozhan
# 筛选 “brand” 列中值为 “Python数据之道” 的行
name = '中国'
df.query('国家奥委会 == @name')
df.query('银牌数 > 30')
df.query('银牌数 > 铜牌数')
# 当需要在某列中筛选多个符合要求的值的时候，可以通过列表（list）来实现   注意里面的单引号和双引号
df.query('国家奥委会 in ["中国","美国"]')
# 
df.query('国家奥委会 in ["中国","美国","英国"] & 银牌数 > 5')
```
