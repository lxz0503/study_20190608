###  8 读取 Excel 文件｜指定列（匹配）



**根据指定列名匹配读取**

让我们来个更难一点的，还是读取 `某招聘网站数据.csv` 文件，但现在有一个 list 中包含多个字段👇

`usecols = ['positionId','test','positionName', 'test1','salary']`

如果 `usecols` 中的列名存在于 `某招聘网站数据.csv` 中，则读取。

```python
# 方法1  利用集合的交集 方法比较笨 但是很直观
data = pd.read_csv('某招聘网站数据.csv')
data.columns
cols = set(['positionId','test','positionName', 'test1','salary'])
# print(cols)
data_cols = set(data.columns)
data_cols
common_cols = list(cols & data_cols)   # 求两个集合的交集  然后转化为列表
# print(common_cols)
data[common_cols].head()   # 求出pandas中对应的列，里面的common_cols是一个列表

# 方法2  作者的方法  也是上面方法的简化版  usecols这个参数就表示要选取其中的列--所以lambda中的c就代表数据中的每一列
cols = ['positionId', 'test', 'positionName', 'test1', 'salary']
pd.read_csv('某招聘网站数据.csv', usecols=lambda c: c in set(cols)).head()
```

### 24 循环读取数据



在本小节 `demodata` 文件夹下有多个 `Excel` 文件，要求一次性循环读取全部文件

```python
import os
path = 'demodata/'
filesnames = os.listdir(path)
filesnames = [f for f in filesnames if f.lower().endswith(".xlsx")]
df_list = []
for filename in filesnames:
    df_list.append(pd.read_excel(path + filename))
df_list
# 下面这行代码设置 ignore_index=True  表示index自然增加，取消原来的index
# pd.concat([df1,df2,df3],ignore_index=True)
# pd.concat适用于 列名完全相同的DataFrame合并  pd.merge适合列名不完全相同的合并
pd.concat(df_list,ignore_index=False)   # False代表不取消原来DataFrame的index
```
