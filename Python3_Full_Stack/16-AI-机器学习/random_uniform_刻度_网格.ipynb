```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
import random

# 可以考虑把performance的数据做成这种折线，显示每天的数据变化  一个月内 
# 生成数据
x = range(60)
y_beijing = [random.uniform(10,15) for i in x] 
y_shanghai = [random.uniform(20,25) for i in x]

# 显示中文 
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 画布 和画图 绘制多条折线图 颜色有 r b g c m y k
plt.figure(figsize=(20,8),dpi=100)
# plt.plot(x,y_beijing,'r*-',label='北京')  # 下面是r * - 各自代表的参数
plt.plot(x,y_beijing,label='北京', color='r',marker='*',linestyle='-') 
plt.plot(x,y_shanghai,'g.-',label='上海')   # marker=.-  或者marker=.

# 刻度
y_ticks = range(40)
x_ticks_labels = ['11:{}分'.format(i) for i in x]

# x y 轴刻度，注意第一个参数必须是数字
plt.yticks(y_ticks[::5])   # 从0开始  每5个取一个数
plt.xticks(x[::5],x_ticks_labels[::5])
# plt.xticks(x[::5])

# 添加网格
plt.grid(True,color='red',ls='-',lw=0.3)   # ls可以设置为. 或者 -  或者-.

# 视频16-3-3-18 添加描述信息
plt.xlabel('时 间',fontdict={'fontsize':15,'color':'b'})
plt.ylabel('温 度',fontdict={'fontsize':15,'color':'b'})
plt.title('一小时温度变化', fontdict={'fontsize':20, 'color':'b'})

# 显示图例来区分不同折线
# 图例可以设置参数 来决定显示位置  默认是best,其它包括  upper right,upper left,lower left,lower right
plt.legend(loc='upper left')  # 里面可以不传任何参数

# 展示图像
plt.show()
```
