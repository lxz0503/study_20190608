```python
import pandas as pd

stock_data = pd.read_csv('stock_day.csv')
```

```python
stock_data.head()
```

```python
stock_data = stock_data.drop(['ma5','ma10','ma20','v_ma5','v_ma10','v_ma20'],axis=1)  # 按列删除一些数据
```

```python
stock_data.head()
```

```python
# pandas直接使用行列索引获取数据   必须是先列后行   shift+enter快捷键运行代码
stock_data['open']['2018-02-27']
```

```python
# loc  只能使用索引名  
stock_data.loc['2018-02-27':'2018-02-14',['open','high']]
```

```python
# 使用iloc可以通过索引的下标取获取数据  前3行  前3列
stock_data.iloc[0:3,0:3]
```

#  16-5-5-3 赋值和排序

```python
# 对某列赋值  data['close']=1 或者data.close=1
# 两种排序  对索引  对内容  ascending=True 升序  默认是升序 False是降序
data = stock_data.sort_values(by=['open','high'],ascending=True).head()
```

```python
data
```

```python
data = stock_data.sort_index(ascending=False).head()   # 降序 按照index排序
```

```python
data
```

```python
# series排序，series可以认为是dataframe中的一列，自然带着index
stock_data['open'].sort_index().head()
```

```python
# series 可以理解为一个key value系列 
cities_prices = {'beijing': 60000, 'shanghai': 59000, 'shenzhen': 58000}
result = pd.Series(cities_prices)   # 把python的字典格式转化为 series，看起来更加直观
result
```

```python
result.values   # 一个列表 
```

```python
result[0],result[1]
```

```python
result.index  # 可以认为是一个列表
```

```python
result.sort_values()   # 默认升序
result.sort_index()
```
