```python
import pandas as pd
import numpy as np

data = pd.read_csv('stock_day.csv')
# data.head()
data['p_change'].head()
```

```python
data_p = data['p_change']
# 数据离散化 采用的API    离散化是用来处理一组连续的数据  分成多个区间 看每个区间多少个数据
qcut_r = pd.qcut(data_p,10)   # 分成10个区间
```

```python
qcut_r.value_counts()    # 查看每个区间多少个数据
```

```python
# 自定义分组 bins
bins = [-100,-7,-5,-3,0,3,5,7,100]
cut_r = pd.cut(data_p,bins=bins)
```

```python
cut_r
```

```python
cut_r.value_counts()    # 查看每个区间多少数据
```

```python
# one hot码  其实就是数据落在哪个区间 然后就对应的设置为1，否则为0
pd.get_dummies(cut_r,prefix='aaa')   # 参数prefix会在每列标题添加
```
