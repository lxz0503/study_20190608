```python
import pandas as pd

left = pd.DataFrame({'key1':['a','b','c','d'],
                     'key2':['c','b','c','d'],
                     'bj':['b','a','c','d'],
                     'sh':['a','b','c','d']
                    })
right = pd.DataFrame({'key1':['a','b','c','d'],
                     'key2':['e','b','c','d'],
                     'tj':['f','b','c','d'],
                     'nj':['a','b','c','d']
                    })
```

```python
left
```

```python
right
```

```python
pd.merge(left,right,on=['key1','key2'])  # 默认是inner join， 都是按列合并，即合并后列数增加 on表示两边都有的列
```

```python
pd.merge(left,right,how='left',on=['key1','key2'])   # left join 以左边的数据为参照 左边不存在的列就是NaN
```

```python
pd.merge(left,right,how='right',on=['key1','key2'])   # right join 以右边的数据为准 右边不存在的列就是NaN
```
