```python
import pandas as pd
import numpy as np

stock_change = np.random.normal(0,1,(10,5))   # 模拟10只股票5天里的涨跌幅  创建符合正态分布的数据
stock_change.all()
```

```python
stock_rise = pd.DataFrame(stock_change)   # 利用DataFrame转化为pandas数据结构
```

```python
stock_rise
```

```python
stock_rise.shape  # 拿到多维数组的shape  是一个元组
```

```python
# 构造行索引
stock_code = ['股票{}'.format(i+1) for i in range(stock_rise.shape[0])]
```

```python
stock_code
```

```python
pd.DataFrame(stock_change,index=stock_code)   # 注意第一个参数的格式，多种多样  这里用了ndarray
```

```python
date = pd.date_range(start='20180501', periods=stock_rise.shape[1],freq='B')   # B表示忽略周末  构造日期列 用函数date_range
```

```python
date
```

```python
stock_c = pd.DataFrame(stock_change,index=stock_code,columns=date)     # 增加列索引的dataframe
```

```python
stock_c
```

```python
stock_c.T   # 行列互换  转置
```
