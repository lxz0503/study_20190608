```python
import pandas as pd
import numpy as np

movie = pd.read_csv('IMDB-Movie-Data.csv')
```

```python
movie.head()
```

```python
np.any(pd.isnull(movie))   # movie里面只要有一个缺失值就返回True
```

```python
data = movie.dropna() # 直接删除缺失值  缺失值类型必须是 np.nan 
```

```python
np.any(data.isnull())   # 表示缺失值已经全部删除
```

```python
# 替换缺失值
movie['Revenue (Millions)'].head()
```

```python
movie['Revenue (Millions)'].fillna(value=movie['Revenue (Millions)'].mean())  # 用平均值替代NaN
```

```python
# 处理所有列的缺失值   前提是每列都是数值类型
for i in movie.columns:
    if np.any(pd.isnull(movie[i])):
        print(i)#
        movie[i].fillna(value=movie[i].mean(),inplace=True)  # 每个缺失值都用平均值来替换
```

```python
np.any(pd.isnull(movie))   # 返回值False表明缺失值全部替换完毕
```
